## 🥉 Exercício 2 — Escalonamento de Chamados: `if-else` + `or`

A central de atendimento decide, para cada chamado, se ele fica com o
atendente padrão ou é **escalonado** para um supervisor. É uma decisão
binária — exatamente dois caminhos, nunca os dois ao mesmo tempo.

**Regra:** escala para supervisor se o tempo de espera passou de **10
minutos** **ou** se o chamado é de um cliente VIP — mesmo que ele tenha
esperado pouco. Os seus dados chegam em `MEU_TEMPO_ESPERA_MIN` e
`MEU_CHAMADO_VIP` (bool).

> 💡 Repare no `or`: **um** dos dois motivos já basta para escalonar. Não é
> preciso os dois ao mesmo tempo — o mesmo tipo de combinação lógica do
> Módulo 6 (`and`/`or`), agora dentro de um `if-else`.

````python
print("=== Central de Atendimento ===")
tempo_espera = MEU_TEMPO_ESPERA_MIN
chamado_vip = MEU_CHAMADO_VIP
LIMITE_ESPERA = 10  # minutos

# TODO 1: um bool, sem if — escala pra supervisor?
escalar_supervisor = chamado_vip or tempo_espera > LIMITE_ESPERA

# TODO 2: if/else — a fila de destino do chamado
if escalar_supervisor == True:
    fila_destino = "🚨 Supervisor"
else:
    fila_destino = "💬 Atendente padrão"

print(f"Tempo de espera: {tempo_espera} min | Cliente VIP: {chamado_vip}")
print(f"Escalar? {escalar_supervisor} -> {fila_destino}")
````

## 🥈 Exercício 4 — Roteador de Entregas: `match`/`case` + `if` aninhado

Cada pedido sai por um **modal** de entrega diferente — moto, bike, van ou
correios — e cada um tem sua própria tarifa de frete-base, prazo e (aqui a
pegadinha) o seu **próprio** valor mínimo para ganhar frete grátis:

| Modal | Frete-base | Prazo | Frete grátis a partir de |
|---|---|---|---|
| `"moto"` | R$ 8,90 | 1 dia | R$ 80 |
| `"bike"` | R$ 5,50 | 1 dia | R$ 60 |
| `"van"` | R$ 15,00 | 2 dias | R$ 300 |
| `"correios"` | R$ 22,00 | 7 dias | R$ 500 |

O seu modal e o valor do seu pedido chegam em `MEU_MODAL_ENTREGA` e
`MEU_VALOR_PEDIDO`. Use `match`/`case` para escolher a linha certa da
tabela e, **dentro de cada `case`** (aninhado), decida se o frete some.

> 💡 Modal desconhecido cai no `case _`, que levanta `ValueError` — mas
> isso não vai acontecer no seu caso, porque o kit só sorteia os quatro
> modais da tabela.

````python
print("=== Roteador de Entregas ===")
modal = MEU_MODAL_ENTREGA
valor_pedido = MEU_VALOR_PEDIDO

# TODO: match/case para frete-base e prazo; DENTRO de cada case, um if/else
# aninhado decide o frete_final usando o mínimo de frete grátis DAQUELE modal.
match modal:
    case "moto":
        frete_base = 8.90
        prazo_dias = 1
        if valor_pedido >= 80.00:
            frete_final = 0.0
        else:
            frete_final = frete_base
    case "bike":
        frete_base = 5.50
        prazo_dias = 1
        if valor_pedido >= 60.00:
            frete_final = 0.0
        else:
            frete_final = frete_base
    case "van":
        frete_base = 15.00
        prazo_dias = 2
        if valor_pedido >= 300.00:
            frete_final = 0.0
        else:
            frete_final = frete_base
    case "correios":
        frete_base = 22.00
        prazo_dias = 7
        if valor_pedido >= 500.00:
            frete_final = 0.0
        else:
            frete_final = frete_base
    case _:
        raise ValueError(f"Modal desconhecido: '{modal}'")

print(f"Modal: {modal} | Pedido: R$ {valor_pedido:.2f}")
print(f"Frete: R$ {frete_final:.2f} | Prazo: {prazo_dias} dia(s)")
````

````python
conferir("exercicio-4", frete_final=frete_final, prazo_dias=prazo_dias)

# TODO: substitua a nota abaixo por uma frase sua sobre este exercício
registrar("exercicio-4", "Vi um vídeo do Hashtag Programação explicando match case e na minha cabeça era mais complicado. Mas depois do vídeo" \
"consegui entender tranquilamente e é bem simples até.")
````

## 🥇 Exercício 5 — Comitê de Expansão: três camadas de `if` aninhado

Você está avaliando se pode abrir uma **segunda unidade** do negócio. O
comitê (leia-se: você mesmo, de forma sistemática) usa três filtros em
**cascata** — cada um só é avaliado se o anterior aprovar:

```
Filtro 1: Porte (faturamento médio trimestral ≥ R$ 60.000)
  └─ aprovado → Filtro 2: Saúde (margem líquida média > 8% e endividamento ≤ 40%)
                  └─ aprovado → Filtro 3: Prontidão (≥ 2 de 3 práticas)
                                   └─ aprovado → ✅ ABRIR segunda unidade
                                   └─ reprovado → ❌ ADIAR EXPANSÃO
                  └─ reprovado → ❌ ADIAR EXPANSÃO
  └─ reprovado → ❌ ADIAR EXPANSÃO
```

Os seus números chegam em `MEU_FATURAMENTO_TRIMESTRE`,
`MINHA_MARGEM_LIQUIDA_MEDIA` e `MEU_ENDIVIDAMENTO_PCT`. As três práticas de
prontidão operacional chegam como bool em `MEU_TEM_SISTEMA_GESTAO`,
`MINHA_CONTABILIDADE_EM_DIA` e `MEU_GERENTE_TREINADO`.

> 💡 `int(True)` é `1` e `int(False)` é `0`. Somar bools conta quantas
> condições são verdadeiras.

````python
print("=== Comitê de Expansão ===")
faturamento = MEU_FATURAMENTO_TRIMESTRE
margem = MINHA_MARGEM_LIQUIDA_MEDIA
endividamento = MEU_ENDIVIDAMENTO_PCT
sistema_gestao = MEU_TEM_SISTEMA_GESTAO
contabilidade_ok = MINHA_CONTABILIDADE_EM_DIA
gerente_treinado = MEU_GERENTE_TREINADO

# TODO 1: conte as práticas de prontidão somando os três bools
praticas_prontidao = contabilidade_ok + sistema_gestao + gerente_treinado

# TODO 2: três filtros aninhados (if dentro de if dentro de if)
# Filtro 1 — Porte
if faturamento >= 60_000:
    # Filtro 2 — Saúde financeira (só chega aqui se o Filtro 1 passou)
    if margem > 0.08 and endividamento <=0.4:
        # Filtro 3 — Prontidão (só chega aqui se o Filtro 2 passou)
        if praticas_prontidao >= 2:
            decisao_expansao = "✅ ABRIR segunda unidade"
        else:
            decisao_expansao = "❌ ADIAR EXPANSÃO"
    else:
        decisao_expansao = "❌ ADIAR EXPANSÃO"
else:
    decisao_expansao = "❌ ADIAR EXPANSÃO"

print(f"Faturamento trimestral: R$ {faturamento:.2f}")
print(f"Margem líquida média: {margem}% | Endividamento: {endividamento}%")
print(f"Prontidão: {praticas_prontidao}/3 práticas")
print(f"Decisão: {decisao_expansao}")
````

````python
conferir(
    "exercicio-5",
    praticas_prontidao=praticas_prontidao,
    decisao_expansao=decisao_expansao,
)

# TODO: substitua a nota abaixo por uma frase sua sobre este exercício
registrar("exercicio-5", "Tenho um pouco mais de dificuldade nessa parte de BOOL mas não sei exatamente porque. Mas consegui fazer sem problemas")
````

O fechamento de caixa do dia processa lançamentos que chegam, de propósito,
**como texto** — exatamente como saem de uma maquininha ou planilha externa:
`MEU_VALOR_CAIXA_TEXTO`, `MINHA_QUANTIDADE_ITENS_TEXTO` e
`MEU_TIPO_LANCAMENTO_TEXTO` (`"entrada"` ou `"saida"`).

Dois tipos de problema podem aparecer:

| Problema | Exemplo | Ferramenta |
|---|---|---|
| Dado ilegível | `"cem reais"`, `""` | `try`/`except ValueError` |
| Regra de negócio violada | valor ou quantidade ≤ 0, tipo inválido | `raise ValueError` |

> 💡 Um único `except ValueError as erro` captura os dois: o que o Python
> levanta sozinho (ex.: `float("cem reais")`) e o que **você** levanta com
> `raise` — porque o seu `raise` produz o mesmo tipo de exceção.

````python
print("=== Fechamento de Caixa ===")
valor_unitario = None
impacto_caixa = None

try:
    # TODO 1: converta os dois campos numéricos
    valor = float(MEU_VALOR_CAIXA_TEXTO)          # float, a partir de MEU_VALOR_CAIXA_TEXTO
    quantidade = int(MINHA_QUANTIDADE_ITENS_TEXTO)      # int, a partir de MINHA_QUANTIDADE_ITENS_TEXTO
    tipo = MEU_TIPO_LANCAMENTO_TEXTO

    # TODO 2: validações de regra de negócio com raise
    if valor <= 0:
        raise ValueError(f"Valor inválido: R$ {valor:.2f} (deve ser positivo)")
    if quantidade <= 0:
        raise ValueError('A quantidade deve ser positiva')   # quantidade deve ser positiva
    if tipo != "entrada" and tipo != "saida":
        raise ValueError("O tipo deve ser entrada ou saída")   # tipo deve ser "entrada" ou "saida"

    # TODO 3: valor unitário e impacto no caixa (entrada soma, saída subtrai)
    valor_unitario = valor / quantidade
    sinal = 1 if tipo == "entrada" else -1
    impacto_caixa = sinal * valor

    print(f"\n✅ Lançamento válido ({tipo})")
    print(f"   Valor total:     R$ {valor:>10,.2f}")
    print(f"   Quantidade:      {quantidade:>10,}")
    print(f"   Valor unitário:  R$ {valor_unitario:>10.2f}")
    print(f"   Impacto no caixa: R$ {impacto_caixa:>9,.2f}")

except ValueError as erro:
    print(f"\n❌ Lançamento DESCARTADO: {erro}")

print("\n--- Pronto para o próximo lançamento. ---")
````

## 🥇 Exercício 7 — Motor de Repasse a Prestadores: `try` + `match` + `if` aninhados

O seu negócio também conecta clientes a prestadores de serviço avulsos
(limpeza, beleza, reparos, aulas) através de um app — e a cada serviço
fechado, o app precisa calcular quanto **repassar** ao prestador.

**Camada 1 — `try`/`except`:** protege o cálculo inteiro contra dado
inválido (defesa, mesmo quando não esperamos que ele falhe).

**Camada 2 — `match`/`case`:** define a **comissão da plataforma** pelo
tipo de serviço, em `MEU_TIPO_SERVICO_PLATAFORMA`:

| Tipo | Comissão da plataforma |
|---|---|
| `"limpeza"` | 20% |
| `"beleza"` | 25% |
| `"reparos"` | 18% |
| `"aulas"` | 15% |

**Camada 3 — `if`/`else` (dentro de cada `case`):** aplica a retenção pelo
regime do prestador, em `MEU_REGIME_PRESTADOR`:

- **`"autonomo"`:** retém **11%** sobre o valor **após** a comissão
  (simplificação didática do INSS).
- **`"mei"`:** sem retenção adicional — só a comissão da plataforma.

Calcule a **comissão**, a **retenção** e o **valor de repasse** final. O
valor do serviço chega em `MEU_VALOR_SERVICO`.

> 💡 **Três níveis de indentação**: `try` → `match` → `if`. Cada nível
> agrega um recuo a mais — e a mesma regrinha do regime se repete dentro de
> **cada** `case`, porque ela vale para todos os tipos de serviço.

````python
print("=== Motor de Repasse a Prestadores ===")
comissao_valor = None
retencao = None
valor_repasse = None

try:
    tipo_servico = MEU_TIPO_SERVICO_PLATAFORMA
    regime = MEU_REGIME_PRESTADOR
    valor_servico = MEU_VALOR_SERVICO

    if valor_servico <= 0:
        raise ValueError(f"Valor de serviço inválido: R$ {valor_servico:.2f}")

    # TODO: match/case para a comissão-base; dentro de cada case, aplique
    # o if/else da retenção pelo regime do prestador.
    match tipo_servico:
        case "limpeza":
            comissao_pct = 0.20   # 20%
            comissao_valor = valor_servico * comissao_pct
            valor_apos_comissao = valor_servico - comissao_valor
            if regime == "autonomo":
                retencao = 0.11* valor_apos_comissao    # 11% do valor_apos_comissao
            else:
                retencao = 0.0
            valor_repasse = valor_apos_comissao - retencao

        case "beleza":
            comissao_pct = 0.25   # 25%
            comissao_valor = valor_servico * comissao_pct
            valor_apos_comissao = valor_servico - comissao_valor
            if regime == "autonomo":
                retencao = 0.11 * valor_apos_comissao
            else:
                retencao = 0.0
            valor_repasse = valor_apos_comissao - retencao

        case "reparos":
            comissao_pct = 0.18   # 18%
            comissao_valor = valor_servico * comissao_pct
            valor_apos_comissao = valor_servico - comissao_valor
            if regime == "autonomo":
                retencao = 0.11 * valor_apos_comissao
            else:
                retencao = 0.0
            valor_repasse = valor_apos_comissao - retencao

        case "aulas":
            comissao_pct = 0.15   # 15%
            comissao_valor = valor_servico * comissao_pct
            valor_apos_comissao = valor_servico - comissao_valor
            if regime == "autonomo":
                retencao = 0.11*valor_apos_comissao
            else:
                retencao = 0.0
            valor_repasse = valor_apos_comissao - retencao

        case _:
            raise ValueError(f"Tipo de serviço desconhecido: '{tipo_servico}'")

    print(f"\n{'='*46}")
    print(f"  REPASSE — {tipo_servico} ({regime})")
    print(f"{'='*46}")
    print(f"  Valor do serviço:  R$ {valor_servico:>10.2f}")
    print(f"  Comissão:          R$ {comissao_valor:>10.2f} ({comissao_pct*100:.0f}%)")
    print(f"  Retenção:          R$ {retencao:>10.2f}")
    print(f"  Repasse final:     R$ {valor_repasse:>10.2f}")
    print(f"{'='*46}")

except ValueError as erro:
    print(f"\n❌ Cálculo interrompido: {erro}")
````

````python
conferir(
    "exercicio-7",
    comissao_valor=comissao_valor,
    retencao=retencao,
    valor_repasse=valor_repasse,
)

# TODO: substitua a nota abaixo por uma frase sua sobre este exercício
registrar("exercicio-7", "Problemas de atençao denovo. Coloquei a porcentagem de aula e beleza errada. Ao invé de colocar o 0.11 tava usando os valores" \
"de porcentagem da tabela por algum motivo")
````
